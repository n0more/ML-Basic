## Задание 2: Подсчет выходных до отпуска

Пользователь вводит сколько дней осталось до ближайшего отпуска. Программа должна вывести количество выходных дней до отпуска, если учесть, что выходные это суббота и воскресенье, сегодня понедельник и праздники мы не учитываем.  

<br>

**Примеры:**
* `4  -> 0`
* `6  -> 1`
* `14 -> 4`

```python
days = 14
weekends = 0
for i in range(days):
    weekday = (i + 0) % 7  
    if weekday in [5, 6]:
        weekends += 1
print(weekends)
```

## Задание 3: Плитка шоколада

Пользователь вводит длину и ширину плитки шоколада, а также размер куска, который хочет отломить. Программа должна вычислить - можно ли совершить подобный разлом или нет, если учесть, что ломать можно только по прямой.

<br>

**Примеры:**
* `3, 4, 6  -> True`
* `5, 7, 8  -> False`
* `4, 5, 12 -> True`

```python
# @title String fields

width = 5 # @param {type:"integer"}
length =  7 # @param {type:"integer"}
chunk = 8 # @param {type:"integer"} 

def f(a, b, d):
     c = a // b
     r = a % b
     if r == 0 and c > 0:
        if c <= d:
            result = True
        else:
            result = False
     else:
        result = False
     return result
    

if chunk % width == 0 :
    result = f(chunk, width, length)
else:
    result = f(chunk, length, width)

print(f"{width}, {length}, {chunk} -> {result}")
```

## Задание 4: Римские числа

Пользователь вводит целое положительное число, программа должна вернуть строку в виде римского числа.

<br>

**Примеры:**
* `3   -> III`
* `15  -> XV`
* `234 -> CCXXXIV`

```python
number = 150 
Roman = []
prev = 0
RomanNumbers = {"I": 1, "V": 5, "X": 10, "L": 50, "C": 100, "D": 500, "M": 1000}
for R in reversed(RomanNumbers):
    o = number // RomanNumbers[R]

    if number == 0:
        break
    # print(f"{RomanNumbers[R] // number} , {number}, {o}, {R}, {RomanNumbers[R]}, {RomanNumbers[R] - RomanNumbers[R]/10}")
    if RomanNumbers[R] // number ==  1 and number >= RomanNumbers[R] - RomanNumbers[R]/10 and o != 1 and prev != 0:
        Roman.append(list(RomanNumbers.keys())[list(RomanNumbers.values()).index(RomanNumbers[R]/10)])
        Roman.append(R)
        number = number - (RomanNumbers[R] - RomanNumbers[R]/10 )
    else:
        if o == 4 and prev != 0:
           Roman.append(R)
           Roman.append(prev)
        else:
            if o != 0:
                for  i in range(o):
                    Roman.append(R)
        number = number - ( o * RomanNumbers[R] )
    prev = R
string = ''.join(Roman)
print(string)
```
