## Задание 1: Получение однозначного числа

Пользователь вводит целое число, программа складывает все цифры числа, с полученным числом — то же самое, и так до тех пор, пока не получится однозначное число.

<br>

**Примеры:**
* `545   -> 5`
* `12345 -> 6`

````python
def sum_digits(number):
    s = 0

    for digit_char in str(number):
        s += int(digit_char)
    return s

def get_single_digit(initial_number):
    if initial_number == 0:
        return 0
    
    current_number = abs(initial_number) 

    while current_number >= 10:
        current_number = sum_digits(current_number)
    
    return current_number

num = 545
result = get_single_digit(num)
print(result)

````

## Задание 2: Кинотеатр

Дан список списков, каждый вложенный список состоит из 1 и 0, количество вложенных списков — количество рядов. Пользователь вводит, сколько билетов ему требуется. Программа должна найти ряд, где можно приобрести нужное количество билетов (места должны быть рядом). Если таких рядов несколько, то ближайший к экрану (ближайшим считается нулевой ряд). Если таких мест нет, то вывести False.

<br>

**Примеры:**
* `[[0,1,1,0], [1,0,0,0], [0,1,0,0]], 2 -> 1`
* `[[0,1,1,0], [1,0,1,0], [1,1,0,1]], 2 -> False`

````python
def find_available_row(cinema_layout, num_tickets):
    
    if num_tickets <= 0:
        print("Количество билетов должно быть положительным числом.")
        return False
    
    for row_index, row in enumerate(cinema_layout):
        current_consecutive_zeros = 0 
        
        for seat in row:
            if seat == 0:
                current_consecutive_zeros += 1 
            else:
                current_consecutive_zeros = 0 
            
            if current_consecutive_zeros >= num_tickets:
                return row_index
                
    return False

cinema_1 = [[0,1,1,0], [1,0,0,0], [0,1,0,0]]
tickets_1 = 2
result_1 = find_available_row(cinema_1, tickets_1)
print(f"Пример 1: {cinema_1}, {tickets_1} -> {result_1}") 

cinema_2 = [[0,1,1,0], [1,0,1,0], [1,1,0,1]]
tickets_2 = 2
result_2 = find_available_row(cinema_2, tickets_2)
print(f"Пример 2: {cinema_2}, {tickets_2} -> {result_2}")
````

## Задание 3: Алгоритм RLE

Необходимо написать упрощенную версию алгоритма RLE. 
Алгоритм RLE объединяет подряд идущие символы в коэффициент и символ.

<br>

**Примеры:**
* `asssdddsssddd -> 1a3s3d3s3d`
* `aaabbbbccccc  -> 3a4b5c`
* `abcba         -> 1a1b1c1b1a`

````python
def simple_rle_encode(input_string):
    if not input_string:
        return ""

    encoded_result = []
    
    current_char = input_string[0]
    current_count = 1

    for i in range(1, len(input_string)):
        if input_string[i] == current_char:
            current_count += 1
        else:
            encoded_result.append(str(current_count) + current_char)
            current_char = input_string[i]
            current_count = 1
    
    encoded_result.append(str(current_count) + current_char)
    
    return "".join(encoded_result)

print(f"'asssdddsssddd' -> {simple_rle_encode('asssdddsssddd')}")
print(f"'aaabbbbccccc'  -> {simple_rle_encode('aaabbbbccccc')}")
````

## Задание 4: Шифр Цезаря

Реализуйте программу для шифрования текста с помощью шифра Цезаря.  
Шифр Цезаря — это метод шифрования, при котором каждая буква в тексте заменяется буквой, стоящей на фиксированное число позиций дальше в алфавите.

### Требования
1. Программа должна принимать на вход:
   - Строку текста для шифрования
   - Ключ шифра (число от 0 до 25)

2. Правила шифрования:
   - Шифруются только латинские буквы (a-z, A-Z)
   - Сохраняется регистр букв (заглавные остаются заглавными)
   - Пробелы, цифры и знаки препинания остаются без изменений
   - Сдвиг циклический (после 'z' идет 'a', после 'Z' идет 'A')

### Примеры

* `"Hello World", 3      -> "Khoor Zruog"`
* `"Python is great!", 5 -> "Udymts nx lwjfy!"`
* `"XYZ abc", 3          -> "ABC def"`
* `"Test 123", 1         -> "Uftu 123"`

### Алгоритм
* Для каждого символа в строке:
   - Если это латинская буква, найти её позицию в алфавите
   - Добавить к позиции значение ключа
   - Если результат больше 25, использовать остаток от деления на 26
   - Преобразовать обратно в букву, сохранив регистр
   - Если это не буква, оставить символ без изменений


````python
def caesar_cipher(text, key):
    result = []
    
    for char in text:
        if 'a' <= char <= 'z':
            start = ord('a')
            shifted_char_code = (ord(char) - start + key) % 26 + start
            result.append(chr(shifted_char_code))
        elif 'A' <= char <= 'Z':
            start = ord('A')
            shifted_char_code = (ord(char) - start + key) % 26 + start
            result.append(chr(shifted_char_code))
        else:
            result.append(char)
            
    return "".join(result)

print(f"\"Hello World\", 3      -> \"{caesar_cipher('Hello World', 3)}\"")
print(f"\"Python is great!\", 5 -> \"{caesar_cipher('Python is great!', 5)}\"")
print(f"\"XYZ abc\", 3          -> \"{caesar_cipher('XYZ abc', 3)}\"")
print(f"\"Test 123\", 1         -> \"{caesar_cipher('Test 123', 1)}\"")
````

## Задание 5: Табель успеваемости

Пользователь в бесконечном цикле (пока не будет введена пустая строка) вводит строки вида: 'название предмета' 'фамилия ученика' 'оценка'. После окончания ввода программа выводит в консоль название предмета, далее список учеников и все их оценки в виде таблицы.

<br>

**Примеры:**

<br>

*Ввод:*
```python
Математика Иванов  5
Математика Иванов  4
Литература Иванов  3
Математика Петров  5
Литература Сидоров 3
Литература Петров  5
Литература Иванов  4
Математика Сидоров 3
Математика Петров  5
```

<br>

*Результат:*
```python
Математика # вводим название предмета

# выводим список учеников и их оценки
Иванов  5 4
Петров  5 5
Сидоров 3
```

```python
Литература # вводим название предмета

# выводим список учеников и их оценки
Иванов  3 4
Сидоров 3
Петров  5
```

````python
def process_grades(input_lines):
    all_grades = {}

    for line in input_lines:
        if not line: # Пустая строка для завершения обработки данных
            break

        parts = line.split()
        if len(parts) != 3:
            continue # Пропускаем некорректные строки

        subject = parts[0]
        student = parts[1]
        
        try:
            grade = int(parts[2])
        except ValueError:
            continue 

        if subject not in all_grades:
            all_grades[subject] = {}
        
        if student not in all_grades[subject]:
            all_grades[subject][student] = []
            
        all_grades[subject][student].append(grade)

    for subject_name, students_data in sorted(all_grades.items()):
        print(subject_name)
        for student_name in sorted(students_data.keys()):
            grades = students_data[student_name]
            grades_str = " ".join(map(str, grades))
            print(f"{student_name}  {grades_str}")
        print()

if __name__ == "__main__":

    example_input_data = """
Математика Иванов  5
Математика Иванов  4
Литература Иванов  3
Математика Петров  5
Литература Сидоров 3
Литература Петров  5
Литература Иванов  4
Математика Сидоров 3
Математика Петров  5
"""

    input_lines_for_processing = example_input_data.strip().split('\n') + ['']

    print("--- Пример ввода ---")
    print(example_input_data)
    print("--- Результат ---")
    process_grades(input_lines_for_processing)
````
