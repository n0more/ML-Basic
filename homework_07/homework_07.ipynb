### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
arr = np.arange(10, 50)

arr = arr[::-1]

even_elements = arr[arr % 2 == 0]

print(arr)
print(even_elements)
```

### Создайте двумерный массив 3x3 со значениями от 0 до 8

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
arr = np.random.rand(4, 3, 2)

min_value = np.min(arr)
max_value = np.max(arr)

print("Массив:")
print(arr)
print(f"Минимум: {min_value}")
print(f"Максимум: {max_value}")
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
array_6x4 = np.random.randint(1, 101, size=(6, 4))
array_4x3 = np.random.randint(1, 101, size=(4, 3))

result = np.dot(array_6x4, array_4x3)

print(array_6x4)
print(array_4x3)
print(result)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
array = np.random.rand(7, 7)

mean = np.mean(array)

std = np.std(array)

normalized_array = (array - mean) / std

print("Исходный массив:")
print(array)
print(f"Среднее значение: {mean}")
print(f"Стандартное отклонение: {std}")
print("Нормализованный массив:")
print(normalized_array)
```

# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
plt.hist(tips['total_bill'], bins=20, edgecolor='black')
plt.title('Распределение total_bill')
plt.xlabel('Сумма total_bill')
plt.ylabel('Количество наблюдений')
plt.show()
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
plt.scatter(tips['total_bill'], tips['tip'])
plt.title('Взаимосвязь между total_bill и tip')
plt.xlabel('Сумма total_bill')
plt.ylabel('Сумма tip')
plt.show()
```

### Постройте pairplot

```python
sns.pairplot(tips)
plt.show()
```

### Постройте график взаимосвязи между признаками 'total_bill' и 'day'

```python
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title('Взаимосвязь между total_bill и day')
plt.xlabel('День')
plt.ylabel('Сумма total_bill')
plt.show()
```

### Постройте две гистограммы распределения признака 'tip' в зависимости от категорий 'time'

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
sns.histplot(tips[tips['time'] == 'Lunch']['tip'], kde=True, color='blue')
plt.title('Распределение tip при Lunch')
plt.xlabel('Сумма tip')
plt.ylabel('Количество наблюдений')

plt.subplot(1, 2, 2)
sns.histplot(tips[tips['time'] == 'Dinner']['tip'], kde=True, color='green')
plt.title('Распределение tip при Dinner')
plt.xlabel('Сумма tip')
plt.ylabel('Количество наблюдений')

plt.tight_layout()
plt.show()
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
sns.scatterplot(x='total_bill', y='tip', hue='smoker', data=tips[tips['sex'] == 'Male'], palette='Set1')
plt.title('Взаимосвязь между total_bill и tip для Male')
plt.xlabel('Сумма total_bill')
plt.ylabel('Сумма tip')

plt.subplot(1, 2, 2)
sns.scatterplot(x='total_bill', y='tip', hue='smoker', data=tips[tips['sex'] == 'Female'], palette='Set2')
plt.title('Взаимосвязь между total_bill и tip для Female')
plt.xlabel('Сумма total_bill')
plt.ylabel('Сумма tip')

plt.tight_layout()
plt.show()
```
